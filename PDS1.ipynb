##numpy.linspace

```python
import numpy as np
# Membuat urutan dari 0 hingga 1 dengan 5 elemen

urutan, jarak = np.linspace(0, 1, 5, retstep=True)
print("Urutan:", urutan)
print("Jarak antara elemen:", jarak)
```

```python
#Membuat urutan bilangan bulat dari 0 hingga 10 dengan 4 elemen dan tipe data integer
urutan = np.linspace(0, 10, 4, dtype=int)

print("Urutan:", urutan)
```

```python
#Membuat urutan dengan num=5 tanpa menggunakan endpoint dan dengan menggunakan retstep
a = np.linspace(2.0, 3.0, num=5)
b = np.linspace(2.0, 3.0, num=5, endpoint=False)
c, step = np.linspace(2.0, 3.0, num=5, retstep=True)

# Menampilkan output
print("Urutan a (dengan endpoint):")
print(a)

print("\nUrutan b (tanpa endpoint):")
print(b)

print("\nUrutan c (dengan retstep):")
print(c)

print("\nJarak antara elemen dalam urutan c:")
print(step)
```

#PANDAS

```python
#P1
#Running pandas dan numpy
import pandas as pd
import numpy as np


#Menampilkan array 1-dimensi
ds = pd.Series([2, 4, 6, 8, 10])
print(ds)
```

```python
#P2
data = pd.Series(['a','b','c'])
np1 = np.array(['x','y','z'])
data_np = pd.Series(np1)

print(data)
print(data_np)

#Mengatur indeks pada numpy array data-np di atas
data_np = pd.Series(np1, index=[2,3,4])
print(data_np)
```

```python
#P3 contoh operasi
ds1 = pd.Series([2, 4, 6, 8, 10])
ds2 = pd.Series([1, 3, 5, 7, 9])
dsa = ds1 + ds2  #operasi jumlah
dsb = ds1 - ds2
dsc = ds1 * ds2
dsd = ds1 / ds2
print(dsa)
print(dsb)
print(dsc)
print(dsd)
```

```python
#P4 konver pandas module Series ke python list
import pandas as pd
ds = pd.Series([2, 4, 6, 8, 10])
print("Pandas Series dan type")
print(ds)
print(type(ds))

print("Konver Pandas Series ke Python list")
print(ds.tolist())
print(type(ds.tolist()))
```

```python
#P5 convert a NumPy array to a Pandas series
import numpy as np
import pandas as pd
np_array = np.array([0, 20, 30, 40, 50])
print("NumPy array:")
print(np_array)

new_np = pd.Series(np_array)
print("konversi Pandas series:")
print(new_np)
```
